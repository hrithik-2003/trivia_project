```python
token = []

print("Available voices:")
for modelTokens, title in zip(voices.modelTokens, voices.title):
     if 'Spider-Woman' in title:
        print(f"Title {title}: {modelTokens}")
        token.append(modelTokens)
```

```python
result = []
for i in token:
    result.append(fy.say("Fuck this World, Fuck the economy, Fuck the market, fuck everything.", tts_model_token=i))
    time.sleep(3)
```

```python

for i,j in enumerate(result):
    print(i)
    with open(f"output/samples/output{i}.wav", "wb") as f:
        f.write(j.content)
    print(f'Voice {i} generated.')
    
```
